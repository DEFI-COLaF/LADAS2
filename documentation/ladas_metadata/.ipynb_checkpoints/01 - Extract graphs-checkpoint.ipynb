```python
import pandas as pd
import glob
import os.path as op
import re
import yaml
import seaborn as sns
import matplotlib.pyplot as plt

gallica = pd.read_excel("./monographies_gallica.xlsx")
persee = pd.read_csv("./metadata_persee_all.csv")
datacatal = pd.read_csv("./catalogs_subset_stats (1).csv")
```

```python
persee["identifier"].iloc[1]
persee["sub"] = persee["identifier"].apply(lambda x: x.split("/")[-1]) 
persee
```

```python

re_gallica = re.compile(r"(b[dp]+[A_Za-z0-9]+)_[f]?(\d+)_jp[e]?g")

data = [
    
]

with open("data/data.yaml") as f:
    config = yaml.safe_load(f)
names = config["names"]

is_gallica = 0

for file in glob.glob("./*/*/images/*"):
    base, _ = op.splitext(op.basename(file))
    dirname = op.dirname(file)
    txt = op.join(dirname.replace("images", "labels"), f"{base}.txt")
    if re_gallica.match(base):
        text_id, page = re_gallica.search(base).groups()
        page = int(page)
        entry = gallica[gallica.URL == f"https://gallica.bnf.fr/ark:/12148/{text_id}"].iloc[0].to_dict()
        if entry:
            is_gallica += 1
        with open(txt) as f:
            for line in f:
                try:
                    idx, x, y, w, h = line.strip().split()
                    idx = int(idx)
                    x, y, w, h = tuple(map(float, [x, y, w, h]))
                    data.append({
                        "id": text_id,
                        "page": page,
                        "pages": entry["NOMBRE DE VUES"],
                        "percent": page / entry["NOMBRE DE VUES"],
                        "date": int(entry["DATES"].split("-")[0]) if isinstance(entry["DATES"], str) else entry["DATES"],
                        "provider": "gallica",
                        "zone": names[idx],
                        "w": w,
                        "h": h
                    })
                except ValueError:
                    continue # print(line.strip())
    elif base.startswith("PG0"):
        
        continue
    else:
        continue # print(base)
    
```

```python
is_gallica
```

```python
df = pd.DataFrame(data)
```

```python
df.drop_duplicates("id").percent.plot.box()
```

```python
plt.figure(figsize=(10, 6))
medians = df.groupby(['zone'])['w'].median().sort_values().index
ax = sns.boxplot(data=df, y="w", x="zone", order=medians)

axis = plt.xticks(rotation=90)  # Adjust the rotation angle as needed
```

```python
fig = plt.figure(figsize=(12, 8), dpi=300)


df["25years"] = df["date"] // 25 * 25
df["10years"] = df["date"] // 10 * 10
ax = sns.histplot(
    x='date',
    y='zone',
    #hue=0,
    #size=0,
    data=df[df['date']>=1500][["date", "zone"]],
    cbar=True,
    #palette="dark:#5A9_r",
    #ax=fig.gca()
)
ax.set_xlabel("Year")
ax.set_ylabel("Zone Type")
# df
plt.tight_layout()
plt.savefig("zoneperyear.png")
```

```python
datas = (df[df['10years']>1500][["10years", "zone"]]
         .groupby("10years")
         .value_counts()
         .reset_index()
         .sort_values(["zone", "10years"])
         .groupby("10years")
         .sum()["count"]
         .reset_index()
         .sort_values(["10years"])
        )
datas.columns = ["year", "zones" ]


fig = plt.figure(dpi=300)
ax = sns.scatterplot(data=datas, x="year", y="zones")
ax.set_xlabel("Decades")
ax.set_ylabel("Number of zones")
```
